```python
#IMPORT STATEMENTS
import numpy as np
import pandas as pd
```

```python
#DEFINITION OF STATE SPACE

#CONFIG
states="ABCDEFGHIJKLMNO"
V=[i for i in states]
start_node="A"
goal_node="O"
#ANALYSIS OF VERTICES
n_nodes=len(V)
print(f"Number of nodes: {n_nodes}\nStart node: {start_node}\nGoal node: {goal_node}\nSize of adjacency matrix will be {n_nodes}x{n_nodes}")
print(f"Initializing a black adjacency matrix of size {n_nodes}x{n_nodes}")
adjacency_matrix = np.zeros((n_nodes,n_nodes),dtype=int)
print(adjacency_matrix)
VERTEX_TO_ID={}
ID_TO_VERTEX={}
for i in range(n_nodes):
  VERTEX_TO_ID[states[i]]=i
  ID_TO_VERTEX[i]=states[i]
print("VERTEX_TO_ID :",VERTEX_TO_ID)
print("ID_TO_VERTEX :",ID_TO_VERTEX)
```

```python
def edge_assignment(start_node,destination_node,weight):
  start_node_id=VERTEX_TO_ID[start_node]
  destination_node_id=VERTEX_TO_ID[destination_node]
  #BUILDING OFF ASSUMPTION THAT EDGES ARE BI-DIRECTIONAL
  adjacency_matrix[start_node_id][destination_node_id]=weight
  adjacency_matrix[destination_node_id][start_node_id]=weight
```

```python
#EDGE ASSIGNMENT PROCESS
print(f"BEFORE:\n{adjacency_matrix}")
edge_assignment("A","B",10)
edge_assignment("A","G",20)
edge_assignment("A","C",10)
edge_assignment("G","F",5)
edge_assignment("G","E",10)
edge_assignment("C","D",30)
edge_assignment("D","H",20)
edge_assignment("H","I",10)
edge_assignment("I","J",15)
edge_assignment("H","N",10)
edge_assignment("N","M",15)
edge_assignment("M","O",20)
edge_assignment("J","L",7)
edge_assignment("I","K",5)
print(f"AFTER:\n{adjacency_matrix}")
```

```python
# REMINDER
# states="ABCDEFGHIJKLMNO"
# V=[i for i in states]
# start_node="A"
# goal_node="O"

#Initializing a visited dictionary for each node with 0 values
visited={}
for i in V:
  visited[i]=0
print(f"Initialised dictionary :{visited}")

#Visiting the start node
stack = []
stack.append(start_node)
visited[start_node]=1
print(f"After start node added: {visited}")
```

###Traversing through the graph

```python
while stack!=[]:
  item=stack[-1]
  item_id=VERTEX_TO_ID[item]
  item_id_weights=adjacency_matrix[item_id]
  options_id=[]
  print(f"Selected: {item}\nExploring {item}...")
  for i in range(n_nodes):
    if item_id_weights[i]!=0:
      options_id.append(i)
  # print(item_id_weights)
  # print(options_id)
  options=[]
  if options_id!=[]:
    for op_id in options_id:
      vertex=ID_TO_VERTEX[op_id]
      options.append(vertex)
  print(f"Possible options: {options}")
  unvisited_options=[op for op in options if visited[op]!=1]
  print(f"Unvisited options: {unvisited_options}")
  if unvisited_options!=[]:
    option=unvisited_options[0]
    print(f"Exploring: {option}")
    if visited[option]==1:
      print(f"{option} already visited..")
    elif visited[option]==0:
      print(f"{option} not visited, adding to stack....")
      visited[option]=1
      stack.append(option)
      print(f"\n\nCurrent stack status: {stack}\n")
    else:
      print(f"Some error happened\n Visited: {visited}")
  else:
    print(f"No further nodes...")
    deleted=stack.pop()
    print(f"Popped {deleted} from stack\n\nCurrent stack:{stack}\n")
```
